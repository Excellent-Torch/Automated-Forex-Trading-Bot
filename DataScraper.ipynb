```python
import asyncio
import pandas as pd
from twscrape import API, gather
from twscrape.logger import set_log_level
```

```python
tweetId = []
userNames = []
rawContent = []

async def ScrapeData(content,no,search_tab=None,type=None):
    api = API()  # or API("path-to.db") - default is `accounts.db`

    # API USAGE
    await api.pool.login_all()

    # change search tab (product), can be: Top, Latest (default), Media
    #await gather(api.search("USA News", limit=20, kv={"currency": "Top"}))


    # change log level, default info
    set_log_level("DEBUG")

    async for tweet in api.search(content, limit=no, kv={search_tab: type}):
        #df = pd.DataFrame({"id": [tweet.id], "user": [tweet.user.username], "rawContent": [tweet.rawContent]})
        
        #print(tweet)
        #tweet_list= [tweet.id[i], tweet.user.username[i], tweet.rawContent[i]]
        #i = i + 1
        #tweets_df=pd.DataFrame(tweet_list, columns=["TweetID","UserName","Tweet_Text"])
        #return tweets_df
        #print(tweet.id, tweet.user.username, tweet.rawContent)
        tweetId.append(tweet.id)
        userNames.append(tweet.user.username)
        rawContent.append(tweet.rawContent)
    
```

```python
def RunScraper():
    
    tweetId.clear() 
    userNames.clear()
    rawContent.clear()

    loop = asyncio.get_event_loop()
    loop.run_until_complete(ScrapeData(CONTENT,NO_TWEETS,SEARCH_TAB,TYPE))

    return True
```

```python
def CreateDataframe():
    RunScraper()
    if RunScraper():
        df = pd.DataFrame(
    {'TweetID': tweetId,
     'UserName': userNames,
     'Tweet_Text': rawContent
    })

    return df
```

```python
CONTENT = 'USA NEWS'
NO_TWEETS = 20
SEARCH_TAB = None #Product
TYPE = None #Top, Latest, Media
```

```python
tweets_df = CreateDataframe()
```
